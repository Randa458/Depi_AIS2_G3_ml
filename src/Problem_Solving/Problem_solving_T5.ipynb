## 151. Reverse Words in a String

```python
def reverse_str(s):
    s = s.split()[::-1]
    new = " ".join(s)
    return new
```

```python
s = "  hello world  "
reverse_str(s)
```

```python
def product_arr(nums):
    zero_count = nums.count(0)
    
    # If more than one zero → everything is zero
    if zero_count > 1:
        return [0] * len(nums)
    
    # Compute product of all non-zero values
    total = 1
    for n in nums:
        if n != 0:
            total *= n
    
    result = []
    for n in nums:
        if zero_count == 0:
            # normal case (no zero): use division
            result.append(total // n)
        else:
            # exactly one zero: only the zero position gets the product
            result.append(total if n == 0 else 0)
    
    return result
```

```python
nums = [-1,-1,0,1,3]
product_arr(nums)
```
