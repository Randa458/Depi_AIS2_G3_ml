```python
class Solution:
    """
    Solution class provides a method to find two indices of numbers
    in a list such that their sum equals a given target.
    """

    def twoSum(self, nums, target):
        """
        Finds two distinct indices in the list where the sum of the
        corresponding numbers equals the target value.

        The function assumes that exactly one solution exists,
        and the same element cannot be used twice.

        Parameters:
            nums (list[int]): A list of integers.
            target (int): The target sum.

        Returns:
            list[int]: A list containing the two indices of the numbers
                       whose sum equals the target.
        """

        # Dictionary to store numbers and their indices
        seen = {}

        # Iterate over the list with index and value
        for i, num in enumerate(nums):

            # Calculate the number needed to reach the target
            complement = target - num

            # Check if the needed number was seen before
            if complement in seen:
                return [seen[complement], i]

            # Store the current number with its index
            seen[num] = i
```

### 2. Add Two Numbers 

```python
class Solution:
    """
    Solution class provides a method to add two numbers represented
    as linked lists.

    Each linked list stores digits in reverse order, where each node
    contains a single digit.
    
    """

    def addTwoNumbers(self, l1, l2):
        """
        Adds two numbers represented by two singly linked lists.

        The digits are stored in reverse order, and each node contains
        a single digit. The function returns the sum as a linked list
        in the same reverse order format.

        Parameters:
            l1 (ListNode): Head of the first linked list.
            l2 (ListNode): Head of the second linked list.

        Returns:
            ListNode: Head of the resulting linked list that represents
                      the sum of the two numbers.
        """

        # Dummy node to simplify linked list construction
        dummy = ListNode(0)
        current = dummy
        carry = 0

        # Loop until both lists and carry are fully processed
        while l1 or l2 or carry:

            # Get current digit values (use 0 if node is None)
            val1 = l1.val if l1 else 0
            val2 = l2.val if l2 else 0

            # Calculate total and carry
            total = val1 + val2 + carry
            carry = total // 10

            # Create a new node with the current digit
            current.next = ListNode(total % 10)
            current = current.next

            # Move to the next nodes if available
            if l1:
                l1 = l1.next
            if l2:
                l2 = l2.next

        # Return the head of the resulting linked list
        return dummy.next
```
