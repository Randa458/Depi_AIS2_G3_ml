### 1. Two Sum

```python
class TwoSum:
    def __init__(self , array : list , target : int ):
        self.array = array
        self.target = target
        
    def two_sum(self):
        for i in range (len(self.array)):
            for j in range (i+1 , len(self.array)):
                if self.array[i] + self. array[j] == self.target:
                    return i , j
            
        return None      

            
    
ts = TwoSum([2, 7, 11, 15], 9)
print(ts.two_sum())
       
            
    
```

### 2. Add Two Numbers 

```python
class Solution:
    """
    This class provides a solution for adding two numbers
    represented by linked lists.
    """

    def addTwoNumbers(self, l1, l2):
        """
        Adds two numbers represented as reversed linked lists.

        Parameters:
            l1 (ListNode): First number as a linked list.
            l2 (ListNode): Second number as a linked list.

        Returns:
            ListNode: Sum of the two numbers as a linked list.
        """

        dummy = ListNode(0)
        current = dummy
        carry = 0

        while l1 or l2 or carry:
            val1 = l1.val if l1 else 0
            val2 = l2.val if l2 else 0

            total = val1 + val2 + carry
            carry = total // 10

            current.next = ListNode(total % 10)
            current = current.next

            if l1:
                l1 = l1.next
            if l2:
                l2 = l2.next

        return dummy.next
```
