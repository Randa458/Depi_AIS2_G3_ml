```python
class Student:
    
    def __init__(self , name , age ): # intialization parameters.
        self.name = name
        self.age = age
        
    def get_info(self):
        return f"my name is {self.name} , and my age is {self.age}."
```

```python
s1 = Student("Randa" , 22)
s1.get_info()
```

```python
from abc import ABC , abstractmethod

class Animal(ABC):
    @abstractmethod
    def move(self):
        pass


class Bird(Animal):
     def move(self):
        print("Move from bird")
        
    
class Cat(Animal):
    def move(self):
        print("Move from cat")
```

```python
a = Bird()
# a.move()
```

```python
b = Animal()
# b.move()
```

```python
class Calculator1:
    def summation(self , n1 , n2):
        return n1 + n2
    
    def abstraction(self , n1 , n2):
        return n1 - n2
        
    def multiplication(self , n1 , n2):
        return n1 * n2
        
    def division(self , n1 , n2):
        return n1 / n2
    
class Calculator2(Calculator1):
    def power(self , n1 , n2):
        return n1 ** n2
        
c1 = Calculator2()
print(c1.summation(2,5))
```

```python
from prettytable import PrettyTable

products = [
    {"name": "Water", "price": 80.0, "quantity": 1500},
    {"name": "Soda", "price": 130.0, "quantity": 1500},
    {"name": "Chips", "price": 75.0, "quantity": 1500},
    {"name": "Bread", "price": 45.0, "quantity": 1500},
    {"name": "Eggs", "price": 200.0, "quantity": 1500},
]

table = PrettyTable()
table.field_names = ["Product Name", "Price ($)", "Quantity"]

for p in products:
    table.add_row([p["name"], f"{p['price']:.2f}", p["quantity"]])

print(table)

total_bill = 0

while True:
    product_name = input("\nEnter product name (or 'exit' to finish): ").strip().lower()
    if product_name == 'exit':
        break

    product = next((p for p in products if p['name'].lower() == product_name), None)

    if product is None:
        print("Product not found. Please try again.")
        continue

    try:
        quantity_required = float(input(f"How many {product['name']} do you want? "))
        if quantity_required <= 0:
            print("Quantity must be greater than zero.")
            continue
        if quantity_required > product["quantity"]:
            print("Insufficient quantity!")
            continue

        discount_percentage = 0.05
        price_before_discount = quantity_required * product['price']
        discount_amount = price_before_discount * discount_percentage
        final_price = price_before_discount - discount_amount

        product['quantity'] -= quantity_required
        total_bill += final_price

        print(f"Success! price for {product['name']}: ${final_price:.2f} (saved: ${discount_amount:.2f})")

        another = input("Something else? (y/n): ").lower()
        if another != "y":
            break

    except ValueError:
        print("Invalid input! Please enter a number.")

print(f"\nTotal Amount to pay: ${total_bill:.2f}")
print("Thank you!")
```
